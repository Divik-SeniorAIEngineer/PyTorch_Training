## Tensor Datatypes


In PyTorch and deep learning, 32-bit floating point (float32) is more precise, while 16-bit floating point (float16) uses less memory and can make computations faster on supported hardware.


## Difference between 32-bit and 16-bit floating point
32-bit
4 bytes

torch.float32
Higher precision
Better for numerical accuracy, model training that is sensitive to rounding, and general-purpose computation.



16-bit
2 bytes

torch.float16
Lower memory use
Uses half the storage per value and can improve throughput on supported GPUs, but has less precision and a narrower numerical range.



precision tells how much detail stored in a single memory

### Multiplication Errors: Broadcasting (shape errors)

- There are two main rules performing matrix multiplication needs to satisfy

  - The **inner dimersions** must be match
  - (3 , 2) @ (3,2) = won't work
  - (2, 3) @ (3,2) = work
  - (3 , 2) @ (2 , 3) = work
- The resulting matrix has the shape of the **outer dimension**
  - (3 , 2) @ (2 , 3) = (3,3)
  - (3 , 2) @ (2 , 4) = (3,4)


```python
tensor @ tensor
```

## PyTorch Reproducability

- Neural Network Training
  - Start with neural numbers
  - Perform tensor operations
  - Update random numbers
- To reduce the randomness in neural networks and PyTorch comes the concept of a **random seed**


```python
torch.rand(3,4)
```

```python
# Using random seed

RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
